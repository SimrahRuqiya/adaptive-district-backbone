```python
import pandas as pd
from pathlib import Path

ASHRAE_DIR = Path("../data/raw/ashrae")

ROOM_CAPACITIES = {
    1: 84,
    2: 32,
    3: 32
}

SENSOR_CONFIG = {
    "co2": {
        "tolerance": 5,
        "column": "co2"
    },
    "temperature": {
        "tolerance": 10,
        "column": "temperature"
    },
    "humidity": {
        "tolerance": 5,
        "column": "humidity"
    },
    "Illuminance": {
        "tolerance": 5,
        "column": "illuminance"
    }
}


def prepare_ashrae_room(room):

    # Load occupancy
    occupancy = pd.read_csv(
        ASHRAE_DIR / f"occupant_count_room_{room}.csv"
    )

    occupancy["time"] = pd.to_timedelta(
        occupancy["Time"]
    )

    occupancy = occupancy.rename(
        columns={f"room_{room}": "occupant_count"}
    )

    daily_results = []

    # Process each recording day separately
    for day_id, day_occupancy in occupancy.groupby("DayId"):

        # Create 5-minute timestamps for this recording day
        grid = pd.DataFrame({
            "time": pd.timedelta_range(
                start="0min",
                periods=288,
                freq="5min"
            )
        })

        grid["DayId"] = day_id

        # Match occupancy to exact 5-minute timestamps
        day_data = grid.merge(
            day_occupancy[["time", "occupant_count"]],
            on="time",
            how="left"
        )

        # Align environmental sensors
        for sensor, config in SENSOR_CONFIG.items():

            sensor_df = pd.read_csv(
                ASHRAE_DIR / f"{sensor}_room_{room}.csv"
            )

            sensor_df["time"] = pd.to_timedelta(
                sensor_df["Time"]
            )

            sensor_day = (
                sensor_df[sensor_df["DayId"] == day_id]
                [["time", f"room_{room}"]]
                .rename(columns={
                    f"room_{room}": config["column"]
                })
                .sort_values("time")
            )

            aligned = pd.merge_asof(
                grid[["time"]].sort_values("time"),
                sensor_day,
                on="time",
                direction="backward",
                tolerance=pd.Timedelta(
                    minutes=config["tolerance"]
                )
            )

            day_data[config["column"]] = aligned[
                config["column"]
            ].values

        day_data["room"] = room
        day_data["room_capacity"] = ROOM_CAPACITIES[room]

        day_data["occupancy_ratio"] = (
            day_data["occupant_count"]
            / day_data["room_capacity"]
        )

        daily_results.append(day_data)

    return pd.concat(daily_results, ignore_index=True)
```

```python
ashrae_rooms = []

for room in [1, 2, 3]:

    room_data = prepare_ashrae_room(room)

    ashrae_rooms.append(room_data)

ashrae_5min = pd.concat(
    ashrae_rooms,
    ignore_index=True
)

print("Dataset shape:", ashrae_5min.shape)

display(ashrae_5min.head())
```

```python
tolerances = [5, 10, 15, 30]

coverage_results = []

for sensor in ["co2", "temperature", "humidity", "Illuminance"]:

    for room in [1, 2, 3]:

        sensor_df = pd.read_csv(
            ASHRAE_DIR / f"{sensor}_room_{room}.csv"
        )

        sensor_df["time"] = pd.to_timedelta(
            sensor_df["Time"]
        )

        for tolerance in tolerances:

            total = 0
            matched = 0

            for day_id in range(44):

                sensor_day = (
                    sensor_df[sensor_df["DayId"] == day_id]
                    [["time"]]
                    .rename(columns={"time": "reading_time"})
                    .sort_values("reading_time")
                )

                grid = pd.DataFrame({
                    "time": pd.timedelta_range(
                        start="0min",
                        periods=288,
                        freq="5min"
                    )
                })

                aligned = pd.merge_asof(
                    grid,
                    sensor_day,
                    left_on="time",
                    right_on="reading_time",
                    direction="backward",
                    tolerance=pd.Timedelta(minutes=tolerance)
                )

                total += len(aligned)
                matched += aligned["reading_time"].notna().sum()

            coverage_results.append({
                "sensor": sensor,
                "room": room,
                "tolerance_min": tolerance,
                "coverage_pct": matched / total * 100
            })

coverage_df = pd.DataFrame(coverage_results)

display(
    coverage_df.pivot_table(
        index=["sensor", "room"],
        columns="tolerance_min",
        values="coverage_pct"
    ).round(2)
)
```

```python
# Check age of matched CO2 readings

age_results = []

for room in [1, 2, 3]:

    sensor_df = pd.read_csv(
        ASHRAE_DIR / f"co2_room_{room}.csv"
    )

    sensor_df["time"] = pd.to_timedelta(
        sensor_df["Time"]
    )

    for day_id in range(44):

        sensor_day = (
            sensor_df[sensor_df["DayId"] == day_id]
            [["time"]]
            .rename(columns={"time": "reading_time"})
            .sort_values("reading_time")
        )

        grid = pd.DataFrame({
            "time": pd.timedelta_range(
                start="0min",
                periods=288,
                freq="5min"
            )
        })

        aligned = pd.merge_asof(
            grid,
            sensor_day,
            left_on="time",
            right_on="reading_time",
            direction="backward",
            tolerance=pd.Timedelta(minutes=5)
        )

        aligned["age_minutes"] = (
            aligned["time"] - aligned["reading_time"]
        ).dt.total_seconds() / 60

        age_results.append(
            aligned[["age_minutes"]].assign(room=room)
        )

age_df = pd.concat(age_results, ignore_index=True)

display(
    age_df.groupby("room")["age_minutes"]
    .describe()
    .round(2)
)
```

```python
import pandas as pd
from pathlib import Path


ROBOD_CAPACITIES = {
    1: 40,
    2: 40,
    3: 15,
    4: 25,
    5: 36
}

ROBOD_ROOM_TYPES = {
    1: "Lecture room",
    2: "Lecture room",
    3: "Office",
    4: "Office",
    5: "Library"
}

robod_frames = []

for room in range(1, 6):

    df = pd.read_csv(
        ROBOD_DIR / f"combined_Room{room}.csv"
    )

    df["timestamp"] = pd.to_datetime(df["timestamp"])

    # Standardize column names
    df = df.rename(columns={
        "occupant_count [number]": "occupant_count",
        "indoor_co2 [ppm]": "co2",
        "air_temperature [Celsius]": "temperature",
        "indoor_relative_humidity [%]": "humidity",
        "illuminance [lux]": "illuminance"
    })

    # Keep common features
    df = df[[
        "timestamp",
        "occupant_count",
        "co2",
        "temperature",
        "humidity",
        "illuminance"
    ]].copy()

    # Add room metadata
    df["source"] = "ROBOD"
    df["room"] = room
    df["room_type"] = ROBOD_ROOM_TYPES[room]
    df["room_capacity"] = ROBOD_CAPACITIES[room]

    # Calculate occupancy ratio
    df["occupancy_ratio"] = (
        df["occupant_count"] / df["room_capacity"]
    )

    # Preserve the original recording date
    df["recording_date"] = df["timestamp"].dt.date

    robod_frames.append(df)

robod_standardized = pd.concat(
    robod_frames,
    ignore_index=True
)

display(robod_standardized.head())

print("Shape:", robod_standardized.shape)
print("\nMissing values:")
display(robod_standardized.isna().sum())

print("\nRows by room:")
display(robod_standardized.groupby("room").size())
```

```python
ashrae_room_types = {
    1: "Lecture room",
    2: "Study zone",
    3: "Study zone"
}

ashrae_standardized = ashrae_5min.copy()

# Add source and room information
ashrae_standardized["source"] = "ASHRAE"

ashrae_standardized["room_type"] = (
    ashrae_standardized["room"].map(ashrae_room_types)
)

# Preserve the original recording session identifier
ashrae_standardized["recording_id"] = (
    "ASHRAE_Day_" +
    ashrae_standardized["DayId"].astype(str)
)

# Preserve time of day
ashrae_standardized["time_of_day"] = (
    ashrae_standardized["time"]
)

# ASHRAE does not provide a complete calendar timestamp
ashrae_standardized["timestamp"] = pd.NA

# Remove the original columns after preserving their information
ashrae_standardized = ashrae_standardized.drop(
    columns=["time", "DayId"]
)

print("ASHRAE standardized shape:", ashrae_standardized.shape)

display(ashrae_standardized.head())
```

```python
robod_final = robod_standardized.copy()

# Create a unique recording identifier for each date and room
robod_final["recording_id"] = (
    "ROBOD_Room_" +
    robod_final["room"].astype(str) +
    "_" +
    robod_final["recording_date"].astype(str)
)

# Extract time of day as elapsed time since midnight
robod_final["time_of_day"] = (
    pd.to_timedelta(
        robod_final["timestamp"].dt.hour * 3600 +
        robod_final["timestamp"].dt.minute * 60 +
        robod_final["timestamp"].dt.second,
        unit="s"
    )
)

# Keep the original timestamp and remove the date helper
robod_final = robod_final.drop(columns=["recording_date"])
```

```python
common_columns = [
    "source",
    "recording_id",
    "timestamp",
    "time_of_day",
    "room",
    "room_type",
    "room_capacity",
    "occupant_count",
    "occupancy_ratio",
    "co2",
    "temperature",
    "humidity",
    "illuminance"
]

ashrae_final = ashrae_standardized[common_columns]
robod_final = robod_final[common_columns]

unified_df = pd.concat(
    [ashrae_final, robod_final],
    ignore_index=True
)

print("Unified dataset shape:", unified_df.shape)

print("\nRows by source:")
display(unified_df["source"].value_counts())

print("\nRows by room:")
display(
    unified_df.groupby(["source", "room"]).size()
)

print("\nMissing values:")
display(unified_df.isna().sum())

display(unified_df.head())
```

```python
robod_final = robod_final.copy()

robod_final["recording_id"] = (
    "ROBOD_" +
    robod_final["timestamp"].dt.strftime("%Y-%m-%d")
)

# Rebuild the unified dataset
unified_df = pd.concat(
    [
        ashrae_standardized[common_columns],
        robod_final[common_columns]
    ],
    ignore_index=True
)

print("Unified shape:", unified_df.shape)

print("\nRecording days by source:")
display(
    unified_df.groupby("source")["recording_id"].nunique()
)
```
